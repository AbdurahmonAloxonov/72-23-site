```python
import sqlite3
from telegram import InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import Application, CommandHandler, CallbackQueryHandler

# SQLite bazasini yaratish
def create_db():
    conn = sqlite3.connect('clicker_game.db')
    cursor = conn.cursor()
    cursor.execute('''
    CREATE TABLE IF NOT EXISTS scores (
        user_id INTEGER PRIMARY KEY,
        username TEXT,
        score INTEGER
    )
    ''')
    conn.commit()
    conn.close()

# Ballarni olish va yangilash
def update_score(user_id, username, increment=1):
    conn = sqlite3.connect('clicker_game.db')
    cursor = conn.cursor()

    cursor.execute('SELECT * FROM scores WHERE user_id = ?', (user_id,))
    user = cursor.fetchone()

    if user:
        cursor.execute('UPDATE scores SET score = score + ? WHERE user_id = ?', (increment, user_id))
    else:
        cursor.execute('INSERT INTO scores (user_id, username, score) VALUES (?, ?, ?)',
                       (user_id, username, increment))

    conn.commit()
    conn.close()

# Ballarni ko'rsatish
async def show_score(update, context):
    user = update.message.from_user
    user_id = user.id
    conn = sqlite3.connect('clicker_game.db')
    cursor = conn.cursor()

    cursor.execute('SELECT score FROM scores WHERE user_id = ?', (user_id,))
    result = cursor.fetchone()
    conn.close()

    if result:
        score = result[0]
        await update.message.reply_text(f"Sizning hozirgi ballaringiz: {score}")
    else:
        await update.message.reply_text("Sizning ballaringiz mavjud emas. O'yinni boshlang!")

# Start komandasi
async def start(update, context):
    user = update.message.from_user
    update_score(user.id, user.username)  # Startda ballarni yaratish

    # Foydalanuvchiga boshlang'ich tugma ko'rsatish
    keyboard = [
        [InlineKeyboardButton("Click to earn coins!", callback_data='click')]
    ]
    reply_markup = InlineKeyboardMarkup(keyboard)
    await update.message.reply_text('Click tugmasini bosing va coin to\'plang!', reply_markup=reply_markup)

# Click tugmasiga bosganda javob berish
async def button(update, context):
    query = update.callback_query
    await query.answer()

    user_id = query.from_user.id
    username = query.from_user.username

    if query.data == 'click':
        update_score(user_id, username, increment=1)  # Har bir clickda 1 coin qo'shish
        await query.edit_message_text(text="Siz coin oldingiz! Yana bosing!")
        keyboard = [
            [InlineKeyboardButton("Click to earn coins!", callback_data='click')]
        ]
        reply_markup = InlineKeyboardMarkup(keyboard)
        await query.message.reply_text("Click tugmasini bosing va coin to'plang!", reply_markup=reply_markup)

# Liderlar jadvalini olish
async def leaderboard(update, context):
    conn = sqlite3.connect('clicker_game.db')
    cursor = conn.cursor()
    cursor.execute('SELECT username, score FROM scores ORDER BY score DESC LIMIT 5')
    leaderboard = cursor.fetchall()
    conn.close()

    leaderboard_text = "Liderlar jadvali:\n"
    if not leaderboard:
        leaderboard_text += "Hozircha hech kimning ballari yo'q."
    else:
        for i, (username, score) in enumerate(leaderboard, 1):
            leaderboard_text += f"{i}. {username} - {score} coins\n"

    await update.message.reply_text(leaderboard_text)

# Asosiy funksiya
async def main():
    # DB yaratish
    create_db()

    # Tokenni o'zgartiring
    application = Application.builder().token("7628124242:AAG_xPyY33hFJabS2jS2xQOmDjD6u-T2P3o").build()

    # /start komandasi uchun handler
    application.add_handler(CommandHandler("start", start))
    application.add_handler(CommandHandler("score", show_score))  # Ballarni ko'rsatish
    application.add_handler(CommandHandler("leaderboard", leaderboard))  # Liderlar jadvali
    application.add_handler(CallbackQueryHandler(button))  # Tugmalarga javob berish

    # Botni ishga tushirish
    await application.run_polling(allowed_updates=["message", "callback_query"])

if __name__ == '__main__':
    import nest_asyncio
    nest_asyncio.apply()  # Google Colab uchun event loopni to'g'ri ishlatish
    import asyncio
    asyncio.run(main())
```
